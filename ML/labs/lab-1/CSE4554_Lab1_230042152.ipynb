### Task 6

```python
numeric_cols = [iris_data[f"f{i}"] for i in range(4)]

print("Info asked for Column-1: ")
print(f"Mean: {np.mean(numeric_cols[0])}")
print(f"Median: {np.median(numeric_cols[0])}")
print(f"Standard Deviation: {np.std(numeric_cols[0])}")

for i, col in enumerate(numeric_cols, start=1):
    print(f"\n--- Column-{i}: Statistical Summary ---")
    print(f"Min:             {np.min(col)}")
    print(f"Max:             {np.max(col)}")
    print(f"Variance:        {np.var(col):.4f}")
    print(f"25th Percentile: {np.percentile(col, 25)}")
    print(f"50th Percentile: {np.percentile(col, 50)}")
    print(f"75th Percentile: {np.percentile(col, 75)}")


```

## Task 2

```python

targeted_indices = df.sample(frac=0.05, random_state=0).index
df.loc[targeted_indices, ["total_bedrooms", "median_income"]] = np.nan

df.loc[targeted_indices].head(n=10)
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
# here the mean is calculated with the non-null values (after modifying 5% rows)
df.fillna(df.mean())
```

## Task 3

```python
correlation = df["median_income"].corr(df["median_house_value"])
print(f"Correlation between INCOME & HOUSE_VALUE: {correlation:.4f}")
```

```python
df["value_to_income_ratio"] = df["median_house_value"] / df["median_income"]
low_value_areas = df.sort_values(by="value_to_income_ratio", ascending=True)

low_value_areas[["latitude", "longitude", "median_income", "median_house_value", "value_to_income_ratio"]].head(10)
```

```python
# choosing with 2 criteria:
#     1. income > mean(income)
#     2. house_value < mean(house_value)
recommended_areas = df[
    (df["median_income"] > df["median_income"].mean()) &
    (df["median_house_value"] < df["median_house_value"].mean())
].sort_values(by="value_to_income_ratio", ascending=True)

# Top-5 recommendations
recommended_areas[["latitude", "longitude", "median_income", "median_house_value", "population", "value_to_income_ratio"]].head(5)
```
