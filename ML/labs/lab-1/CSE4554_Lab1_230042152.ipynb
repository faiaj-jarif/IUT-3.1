```python
def extract_subset(x, y, y0):
  return x[y == y0]

print(extract_subset(x, y, y0=0))
print()
print(extract_subset(x, y, y0=1))
```

### Task 5

```python
iris_data = np.genfromtxt(
    fname="iris.data",
    delimiter=",",
    dtype=None,
    encoding="utf-8"
)

print(iris_data)
```

### Task 6

```python
numeric_cols = [
    iris_data["f0"],
    iris_data["f1"],
    iris_data["f2"],
    iris_data["f3"],
]

mean_val = np.mean(numeric_cols[0])
median_val = np.median(numeric_cols[0])
std_val = np.std(numeric_cols[0])

print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Standard Deviation: {std_val}")

for i, col in enumerate(numeric_cols, start=1):
    print(f"\n--- Column-{i}: Statistical Summary ---")
    print(f"Min:             {np.min(col)}")
    print(f"Max:             {np.max(col)}")
    print(f"Variance:        {np.var(col):.4f}")
    print(f"25th Percentile: {np.percentile(col, 25)}")
    print(f"50th Percentile: {np.percentile(col, 50)}")
    print(f"75th Percentile: {np.percentile(col, 75)}")
```
